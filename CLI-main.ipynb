```python
import json
from datetime import datetime
```

```python
TASKS_FILE = "tasks.json"
```

```python
def load_tasks():
    try:
        with open(TASKS_FILE, "r") as f:
            return json.load(f)
    except FileNotFoundError:
        return []

def save_tasks(tasks):
    with open(TASKS_FILE, "w") as f:
        json.dump(tasks, f, indent=4)

def add_task(description):
    tasks = load_tasks()
    new_task = {
        "id": len(tasks) + 1,
        "description": description,
        "status": "todo",
        "created_at": datetime.now().isoformat(),
        "updated_at": datetime.now().isoformat()
    }
    tasks.append(new_task)
    save_tasks(tasks)
    print(f"Task added: {description}  ID:({len(tasks)})")

def update(id, new_description):
    tasks = load_tasks()
    tasks[id -1].update({"description":new_description})
    tasks[id -1].update({"updated_at":datetime.now().isoformat()})
    save_tasks(tasks)
    print(f"Task {id} updated: {new_description}")

def delete(id):
    tasks = load_tasks()
    tasks.pop(id-1)
    save_tasks(tasks)
    print("Task deleted successfully")

def mark(id, new_status):
    tasks = load_tasks()
    if new_status not in {"todo","in-progress","done"}:
        raise TypeError('Invalid operation')
    tasks[id -1].update({"status":new_status})
    tasks[id -1].update({"updated_at":datetime.now().isoformat()})
    save_tasks(tasks)
    print(f"{tasks[id -1].get("description")} : {new_status}")
    
```

```python
mark(4,'in-progress')
```
